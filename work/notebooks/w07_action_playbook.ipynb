## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
df_ranked = df.sort_values(by="model_score", ascending=False).reset_index( drop=True )
def assign_reason_codes(row):
  codes = []
  if row.get("days_since_last_update", 0) > 365:
    codes.append("OUTDATED_CONTENT")

  if row.get("ctr", 1.0) < 0.02:
    codes.append("LOW_CTR_SNIPPET_ISSUE")

  if row.get("engagement_rate", 1.0) < 0.40:
    codes.append("LOW_USER_ENGAGEMENT")

  if row.get("scroll_rate", 1.0) < 0.30:
    codes.append("POOR_PAGE_RETENTION")

  if row.get("search_volume", 0) > 1000 and row.get("avg_position", 100) > 15:
    codes.append("HIGH_POTENTIAL_LOW_RANK")

  if not codes:
    codes.append("HIGH_MODEL_CONFIDENCE")

  return ", ".join(codes[:2])


df_ranked["reason_codes"] = df_ranked.apply(assign_reason_codes, axis=1)
action_map = {
    "guide": "Full Editorial Rewrite & Fact Update",
    "blog_post": "Title & Meta Snippet Refresh",
    "landing_page": "UX Layout & Call-To-Action Optimization",
    "product": "Pricing & Availability Audit",
    "article": "Internal Links & Citation Boost",
    }

df_ranked["recommended_action"] = df_ranked["content_type"].map( lambda x: action_map.get(str(x).lower(), "General Editorial Review") )

print("--- TOP 10 PRIORITIZED CONTENT REFRESH QUEUE ---")
display( df_ranked[[
    "client_id",
    "content_type",
    "model_score",
    "recommended_action",
    "reason_codes",
    ]].head(10)
    )
```
